```python
# Customer Engagement, Product Utilization and Churn Exploratory Data Analysis

# Import required libraries
import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

warnings.filterwarnings("ignore")

# Configure display and visualization settings
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 160)
pd.set_option("display.float_format", "{:.2f}".format)

sns.set_theme(style="whitegrid", context="notebook")

plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["figure.dpi"] = 100
plt.rcParams["savefig.dpi"] = 150


# Define project folder paths
current_directory = os.getcwd()

project_folder_name = (
    "Customer Engagement & Product Utilization Analytics for Retention Strategy"
)

if os.path.basename(current_directory).lower() == "notebooks":
    project_folder = os.path.dirname(current_directory)
else:
    project_folder = os.path.join(current_directory, project_folder_name)

processed_data_folder = os.path.join(
    project_folder, "data", "processed"
)

output_folder = os.path.join(project_folder, "outputs")

figure_folder = os.path.join(output_folder, "figures")
table_folder = os.path.join(output_folder, "tables")
eda_figure_folder = os.path.join(figure_folder, "eda")

os.makedirs(eda_figure_folder, exist_ok=True)
os.makedirs(table_folder, exist_ok=True)

input_path = os.path.join(
    processed_data_folder,
    "European_Bank_Cleaned_Analytical.csv"
)

print("Project folder:", project_folder)
print("Input dataset:", input_path)


# Load cleaned analytical dataset
if not os.path.exists(input_path):
    raise FileNotFoundError(
        "Cleaned analytical dataset not found. "
        "Please complete Phase 2 successfully before running EDA."
    )

df = pd.read_csv(input_path)

print("\nAnalytical dataset loaded successfully.")
print("Dataset shape:", df.shape)


# Validate required columns
required_columns = [
    "customerid",
    "creditscore",
    "geography",
    "gender",
    "age",
    "tenure",
    "balance",
    "numofproducts",
    "hascrcard",
    "isactivemember",
    "estimatedsalary",
    "exited",
    "age_group",
    "tenure_group",
    "balance_category",
    "product_count_category",
    "activity_status",
    "customer_status",
    "high_balance_flag",
    "inactive_high_balance_flag",
    "premium_customer_candidate",
    "relationship_strength_index",
    "relationship_strength_category"
]

missing_columns = [
    column for column in required_columns
    if column not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Required columns are missing: {missing_columns}"
    )

# Ensure numerical columns have numeric data types
numerical_columns = [
    "creditscore",
    "age",
    "tenure",
    "balance",
    "numofproducts",
    "hascrcard",
    "isactivemember",
    "estimatedsalary",
    "exited",
    "relationship_strength_index"
]

for column in numerical_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )

if df[numerical_columns].isnull().any().any():
    raise ValueError(
        "Missing or non-numeric values detected in required numerical columns."
    )

# Create reusable plotting and saving functions
eda_tables = {}
eda_findings = []

def save_figure(filename):
    """Save the current chart in the EDA figure folder."""
    file_path = os.path.join(
        eda_figure_folder,
        filename
    )
    plt.tight_layout()
    plt.savefig(file_path, bbox_inches="tight")
    plt.show()
    plt.close()
    return file_path


def save_table(table, filename):
    """Save an analytical table as a CSV file."""
    file_path = os.path.join(
        table_folder,
        filename
    )
    table.to_csv(file_path, index=True)
    return file_path


def add_finding(title, observation):
    """Store an analytical observation for the EDA findings report."""
    eda_findings.append({
        "Analysis Area": title,
        "Observation": str(observation)
    })


def plot_count_distribution(column, title, xlabel, filename, rotation=0):
    """Generate a count distribution chart."""
    plt.figure(figsize=(10, 6))

    order = df[column].value_counts().index

    sns.countplot(
        data=df,
        x=column,
        order=order
    )

    plt.title(title, fontsize=14, fontweight="bold")
    plt.xlabel(xlabel)
    plt.ylabel("Customer Count")
    plt.xticks(rotation=rotation)

    save_figure(filename)


def plot_churn_rate(column, title, xlabel, filename, rotation=0):
    """Compare churn rates across customer groups."""
    churn_table = (
        df.groupby(column, observed=False)
        .agg(
            Total_Customers=("exited", "count"),
            Exited_Customers=("exited", "sum")
        )
    )

    churn_table["Churn_Rate_Percentage"] = (
        churn_table["Exited_Customers"] /
        churn_table["Total_Customers"] * 100
    ).round(2)

    eda_tables[f"{column}_churn"] = churn_table

    plt.figure(figsize=(10, 6))

    sns.barplot(
        x=churn_table.index.astype(str),
        y=churn_table["Churn_Rate_Percentage"]
    )

    plt.title(title, fontsize=14, fontweight="bold")
    plt.xlabel(xlabel)
    plt.ylabel("Churn Rate (%)")
    plt.xticks(rotation=rotation)

    save_figure(filename)

    return churn_table


# ============================================================
# UNIVARIATE ANALYSIS
# ============================================================

print("\nUNIVARIATE ANALYSIS")


# Customer distribution
customer_distribution = pd.DataFrame({
    "Total Customers": [len(df)],
    "Unique Customer IDs": [df["customerid"].nunique()]
})

eda_tables["customer_distribution"] = customer_distribution

print("\nCustomer Distribution")
display(customer_distribution)

plt.figure(figsize=(7, 5))
plt.bar(
    ["Total Customers"],
    [len(df)]
)
plt.title("Total Customer Distribution")
plt.ylabel("Customer Count")
save_figure("01_customer_distribution.png")


# Credit score distribution
print("\nCredit Score Distribution")
display(df["creditscore"].describe().to_frame("Credit Score"))

plt.figure(figsize=(10, 6))
sns.histplot(
    data=df,
    x="creditscore",
    bins=30,
    kde=True
)
plt.title("Credit Score Distribution")
plt.xlabel("Credit Score")
plt.ylabel("Customer Count")
save_figure("02_credit_score_distribution.png")


# Age distribution
print("\nAge Distribution")
display(df["age"].describe().to_frame("Age"))

plt.figure(figsize=(10, 6))
sns.histplot(
    data=df,
    x="age",
    bins=30,
    kde=True
)
plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Customer Count")
save_figure("03_age_distribution.png")


# Tenure distribution
print("\nTenure Distribution")
tenure_distribution = df["tenure"].value_counts().sort_index().to_frame(
    "Customer Count"
)
display(tenure_distribution)

eda_tables["tenure_distribution"] = tenure_distribution

plot_count_distribution(
    "tenure",
    "Customer Tenure Distribution",
    "Tenure (Years)",
    "04_tenure_distribution.png"
)


# Balance distribution
print("\nBalance Distribution")
display(df["balance"].describe().to_frame("Balance"))

plt.figure(figsize=(10, 6))
sns.histplot(
    data=df,
    x="balance",
    bins=40,
    kde=True
)
plt.title("Customer Balance Distribution")
plt.xlabel("Balance")
plt.ylabel("Customer Count")
save_figure("05_balance_distribution.png")


# Estimated salary distribution
print("\nEstimated Salary Distribution")
display(df["estimatedsalary"].describe().to_frame("Estimated Salary"))

plt.figure(figsize=(10, 6))
sns.histplot(
    data=df,
    x="estimatedsalary",
    bins=40,
    kde=True
)
plt.title("Estimated Salary Distribution")
plt.xlabel("Estimated Salary")
plt.ylabel("Customer Count")
save_figure("06_estimated_salary_distribution.png")


# Product-count distribution
print("\nProduct Count Distribution")
product_distribution = (
    df["numofproducts"]
    .value_counts()
    .sort_index()
    .to_frame("Customer Count")
)

eda_tables["product_distribution"] = product_distribution

display(product_distribution)

plot_count_distribution(
    "numofproducts",
    "Customer Product Count Distribution",
    "Number of Products",
    "07_product_count_distribution.png"
)


# Activity distribution
print("\nActivity Distribution")
activity_distribution = (
    df["activity_status"]
    .value_counts()
    .to_frame("Customer Count")
)

eda_tables["activity_distribution"] = activity_distribution

display(activity_distribution)

plot_count_distribution(
    "activity_status",
    "Active vs Inactive Customers",
    "Activity Status",
    "08_activity_distribution.png"
)


# Geography distribution
print("\nGeography Distribution")
geography_distribution = (
    df["geography"]
    .value_counts()
    .to_frame("Customer Count")
)

eda_tables["geography_distribution"] = geography_distribution

display(geography_distribution)

plot_count_distribution(
    "geography",
    "Customer Geography Distribution",
    "Geography",
    "09_geography_distribution.png"
)


# Gender distribution
print("\nGender Distribution")
gender_distribution = (
    df["gender"]
    .value_counts()
    .to_frame("Customer Count")
)

eda_tables["gender_distribution"] = gender_distribution

display(gender_distribution)

plot_count_distribution(
    "gender",
    "Customer Gender Distribution",
    "Gender",
    "10_gender_distribution.png"
)


# Credit card ownership distribution
print("\nCredit Card Ownership Distribution")

credit_card_distribution = (
    df["hascrcard"]
    .map({0: "No Credit Card", 1: "Has Credit Card"})
    .value_counts()
    .to_frame("Customer Count")
)

eda_tables["credit_card_distribution"] = credit_card_distribution

display(credit_card_distribution)

df["credit_card_status"] = df["hascrcard"].map({
    0: "No Credit Card",
    1: "Has Credit Card"
})

plot_count_distribution(
    "credit_card_status",
    "Credit Card Ownership Distribution",
    "Credit Card Status",
    "11_credit_card_distribution.png"
)


# Churn distribution
print("\nChurn Distribution")

churn_distribution = (
    df["customer_status"]
    .value_counts()
    .to_frame("Customer Count")
)

churn_distribution["Percentage"] = (
    df["customer_status"]
    .value_counts(normalize=True) * 100
).round(2)

eda_tables["churn_distribution"] = churn_distribution

display(churn_distribution)

plot_count_distribution(
    "customer_status",
    "Customer Churn Distribution",
    "Customer Status",
    "12_churn_distribution.png"
)


# Record overall churn baseline
overall_churn_rate = df["exited"].mean() * 100

add_finding(
    "Overall Churn",
    f"Overall observed churn rate is {overall_churn_rate:.2f}%."
)


# ============================================================
# BIVARIATE ANALYSIS
# ============================================================

print("\nBIVARIATE ANALYSIS")


# Age vs churn
age_churn = (
    df.groupby("age_group", observed=False)
    .agg(
        Total_Customers=("exited", "count"),
        Exited_Customers=("exited", "sum"),
        Average_Age=("age", "mean")
    )
)

age_churn["Churn_Rate_Percentage"] = (
    age_churn["Exited_Customers"] /
    age_churn["Total_Customers"] * 100
).round(2)

eda_tables["age_vs_churn"] = age_churn

print("\nAge vs Churn")
display(age_churn)

plot_churn_rate(
    "age_group",
    "Age Group vs Churn Rate",
    "Age Group",
    "13_age_vs_churn.png"
)


# Geography vs churn
print("\nGeography vs Churn")

geography_churn = plot_churn_rate(
    "geography",
    "Geography vs Churn Rate",
    "Geography",
    "14_geography_vs_churn.png"
)

display(geography_churn)


# Gender vs churn
print("\nGender vs Churn")

gender_churn = plot_churn_rate(
    "gender",
    "Gender vs Churn Rate",
    "Gender",
    "15_gender_vs_churn.png"
)

display(gender_churn)


# Credit score vs churn
print("\nCredit Score vs Churn")

credit_score_churn = (
    df.groupby("exited")
    .agg(
        Average_Credit_Score=("creditscore", "mean"),
        Median_Credit_Score=("creditscore", "median"),
        Customer_Count=("customerid", "count")
    )
)

credit_score_churn.index = credit_score_churn.index.map({
    0: "Retained",
    1: "Exited"
})

eda_tables["credit_score_vs_churn"] = credit_score_churn

display(credit_score_churn)

plt.figure(figsize=(9, 6))
sns.boxplot(
    data=df,
    x="customer_status",
    y="creditscore"
)
plt.title("Credit Score vs Customer Churn")
plt.xlabel("Customer Status")
plt.ylabel("Credit Score")
save_figure("16_credit_score_vs_churn.png")


# Tenure vs churn
print("\nTenure vs Churn")

tenure_churn = plot_churn_rate(
    "tenure_group",
    "Tenure Group vs Churn Rate",
    "Tenure Group",
    "17_tenure_vs_churn.png"
)

display(tenure_churn)


# Balance vs churn
print("\nBalance vs Churn")

balance_churn = (
    df.groupby("customer_status")
    .agg(
        Average_Balance=("balance", "mean"),
        Median_Balance=("balance", "median"),
        Customer_Count=("customerid", "count")
    )
)

eda_tables["balance_vs_churn"] = balance_churn

display(balance_churn)

plt.figure(figsize=(9, 6))
sns.boxplot(
    data=df,
    x="customer_status",
    y="balance"
)
plt.title("Customer Balance vs Churn")
plt.xlabel("Customer Status")
plt.ylabel("Balance")
save_figure("18_balance_vs_churn.png")


# Salary vs churn
print("\nEstimated Salary vs Churn")

salary_churn = (
    df.groupby("customer_status")
    .agg(
        Average_Salary=("estimatedsalary", "mean"),
        Median_Salary=("estimatedsalary", "median"),
        Customer_Count=("customerid", "count")
    )
)

eda_tables["salary_vs_churn"] = salary_churn

display(salary_churn)

plt.figure(figsize=(9, 6))
sns.boxplot(
    data=df,
    x="customer_status",
    y="estimatedsalary"
)
plt.title("Estimated Salary vs Churn")
plt.xlabel("Customer Status")
plt.ylabel("Estimated Salary")
save_figure("19_salary_vs_churn.png")


# Product count vs churn
print("\nProduct Count vs Churn")

product_churn = plot_churn_rate(
    "numofproducts",
    "Product Count vs Churn Rate",
    "Number of Products",
    "20_product_count_vs_churn.png"
)

display(product_churn)


# Activity status vs churn
print("\nActivity Status vs Churn")

activity_churn = plot_churn_rate(
    "activity_status",
    "Activity Status vs Churn Rate",
    "Activity Status",
    "21_activity_vs_churn.png"
)

display(activity_churn)


# Credit card ownership vs churn
print("\nCredit Card Ownership vs Churn")

credit_card_churn = plot_churn_rate(
    "credit_card_status",
    "Credit Card Ownership vs Churn Rate",
    "Credit Card Status",
    "22_credit_card_vs_churn.png"
)

display(credit_card_churn)


# ============================================================
# MULTIVARIATE ANALYSIS
# ============================================================

print("\nMULTIVARIATE ANALYSIS")


# Correlation analysis
print("\nCorrelation Analysis")

correlation_columns = [
    "creditscore",
    "age",
    "tenure",
    "balance",
    "numofproducts",
    "hascrcard",
    "isactivemember",
    "estimatedsalary",
    "exited",
    "high_balance_flag",
    "premium_customer_candidate",
    "relationship_strength_index"
]

correlation_matrix = df[
    correlation_columns
].corr()

eda_tables["correlation_matrix"] = correlation_matrix

display(correlation_matrix)

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title("Customer Feature Correlation Matrix")
save_figure("23_correlation_matrix.png")


# Numerical feature relationship analysis
print("\nNumerical Feature Relationship Analysis")

numerical_relationships = [
    ("age", "creditscore"),
    ("balance", "estimatedsalary"),
    ("age", "balance"),
    ("tenure", "numofproducts")
]

numerical_relationship_results = []

for x_column, y_column in numerical_relationships:

    correlation_value = df[
        [x_column, y_column]
    ].corr().iloc[0, 1]

    numerical_relationship_results.append({
        "Feature 1": x_column,
        "Feature 2": y_column,
        "Correlation": correlation_value
    })

    plt.figure(figsize=(9, 6))

    sns.scatterplot(
        data=df,
        x=x_column,
        y=y_column,
        alpha=0.4
    )

    plt.title(
        f"{x_column.title()} vs {y_column.title()}"
    )

    plt.xlabel(x_column.title())
    plt.ylabel(y_column.title())

    filename = (
        f"24_relationship_{x_column}_{y_column}.png"
    )

    save_figure(filename)

numerical_relationship_df = pd.DataFrame(
    numerical_relationship_results
)

eda_tables["numerical_relationships"] = numerical_relationship_df

display(numerical_relationship_df)


# Activity and product-count interaction
print("\nActivity and Product Count Interaction")

activity_product_table = pd.crosstab(
    df["activity_status"],
    df["numofproducts"],
    values=df["exited"],
    aggfunc="mean"
) * 100

activity_product_table = activity_product_table.round(2)

eda_tables["activity_product_interaction"] = activity_product_table

display(activity_product_table)

plt.figure(figsize=(10, 6))

sns.heatmap(
    activity_product_table,
    annot=True,
    fmt=".2f",
    cmap="YlOrRd"
)

plt.title("Churn Rate: Activity Status vs Product Count")
plt.xlabel("Number of Products")
plt.ylabel("Activity Status")

save_figure("25_activity_product_interaction.png")


# Age and product-count interaction
print("\nAge and Product Count Interaction")

age_product_table = pd.pivot_table(
    df,
    index="age_group",
    columns="numofproducts",
    values="exited",
    aggfunc="mean",
    observed=False
) * 100

age_product_table = age_product_table.round(2)

eda_tables["age_product_interaction"] = age_product_table

display(age_product_table)

plt.figure(figsize=(10, 6))

sns.heatmap(
    age_product_table,
    annot=True,
    fmt=".2f",
    cmap="YlOrRd"
)

plt.title("Churn Rate: Age Group vs Product Count")
plt.xlabel("Number of Products")
plt.ylabel("Age Group")

save_figure("26_age_product_interaction.png")


# Geography and activity interaction
print("\nGeography and Activity Interaction")

geography_activity_table = pd.pivot_table(
    df,
    index="geography",
    columns="activity_status",
    values="exited",
    aggfunc="mean"
) * 100

geography_activity_table = geography_activity_table.round(2)

eda_tables["geography_activity_interaction"] = geography_activity_table

display(geography_activity_table)

plt.figure(figsize=(9, 6))

sns.heatmap(
    geography_activity_table,
    annot=True,
    fmt=".2f",
    cmap="YlOrRd"
)

plt.title("Churn Rate: Geography vs Activity Status")
plt.xlabel("Activity Status")
plt.ylabel("Geography")

save_figure("27_geography_activity_interaction.png")


# Balance and activity interaction
print("\nBalance and Activity Interaction")

balance_activity_table = pd.pivot_table(
    df,
    index="balance_category",
    columns="activity_status",
    values="exited",
    aggfunc="mean",
    observed=False
) * 100

balance_activity_table = balance_activity_table.round(2)

eda_tables["balance_activity_interaction"] = balance_activity_table

display(balance_activity_table)

plt.figure(figsize=(9, 6))

sns.heatmap(
    balance_activity_table,
    annot=True,
    fmt=".2f",
    cmap="YlOrRd"
)

plt.title("Churn Rate: Balance Category vs Activity Status")
plt.xlabel("Activity Status")
plt.ylabel("Balance Category")

save_figure("28_balance_activity_interaction.png")


# Customer profile comparison
print("\nCustomer Profile Comparison")

customer_profile = (
    df.groupby("customer_status")
    .agg(
        Customer_Count=("customerid", "count"),
        Average_Age=("age", "mean"),
        Average_Credit_Score=("creditscore", "mean"),
        Average_Tenure=("tenure", "mean"),
        Average_Balance=("balance", "mean"),
        Average_Products=("numofproducts", "mean"),
        Average_Salary=("estimatedsalary", "mean"),
        Active_Customer_Percentage=("isactivemember", "mean"),
        High_Balance_Percentage=("high_balance_flag", "mean"),
        Premium_Candidate_Percentage=("premium_customer_candidate", "mean"),
        Average_RSI=("relationship_strength_index", "mean")
    )
)

percentage_columns = [
    "Active_Customer_Percentage",
    "High_Balance_Percentage",
    "Premium_Candidate_Percentage"
]

customer_profile[percentage_columns] = (
    customer_profile[percentage_columns] * 100
).round(2)

customer_profile = customer_profile.round(2)

eda_tables["customer_profile_comparison"] = customer_profile

display(customer_profile)

profile_plot_columns = [
    "Average_Age",
    "Average_Balance",
    "Average_Products",
    "Average_RSI"
]

for column in profile_plot_columns:

    plt.figure(figsize=(8, 5))

    sns.barplot(
        data=customer_profile.reset_index(),
        x="customer_status",
        y=column
    )

    plt.title(
        f"{column.replace('_', ' ')} by Customer Status"
    )

    plt.xlabel("Customer Status")
    plt.ylabel(column.replace("_", " "))

    save_figure(
        f"29_customer_profile_{column.lower()}.png"
    )


# ============================================================
# GENERATE EDA INSIGHTS AND SAVE OUTPUTS
# ============================================================

print("\nGENERATING EDA INSIGHTS")


# Identify highest churn age group
if not age_churn.empty:

    highest_age_group = age_churn[
        "Churn_Rate_Percentage"
    ].idxmax()

    highest_age_churn = age_churn[
        "Churn_Rate_Percentage"
    ].max()

    add_finding(
        "Age and Churn",
        f"{highest_age_group} has the highest observed churn rate "
        f"among age groups at {highest_age_churn:.2f}%."
    )


# Identify highest churn geography
if not geography_churn.empty:

    highest_geography = geography_churn[
        "Churn_Rate_Percentage"
    ].idxmax()

    highest_geography_churn = geography_churn[
        "Churn_Rate_Percentage"
    ].max()

    add_finding(
        "Geography and Churn",
        f"{highest_geography} has the highest observed geography-level "
        f"churn rate at {highest_geography_churn:.2f}%."
    )


# Identify activity-related churn difference
if not activity_churn.empty:

    if "Active" in activity_churn.index and "Inactive" in activity_churn.index:

        active_rate = activity_churn.loc[
            "Active", "Churn_Rate_Percentage"
        ]

        inactive_rate = activity_churn.loc[
            "Inactive", "Churn_Rate_Percentage"
        ]

        add_finding(
            "Activity and Churn",
            f"Active customer churn rate is {active_rate:.2f}%, "
            f"while inactive customer churn rate is {inactive_rate:.2f}%."
        )


# Identify product-count churn pattern
if not product_churn.empty:

    highest_product_group = product_churn[
        "Churn_Rate_Percentage"
    ].idxmax()

    highest_product_churn = product_churn[
        "Churn_Rate_Percentage"
    ].max()

    add_finding(
        "Product Adoption and Churn",
        f"Customers with {highest_product_group} products have the "
        f"highest observed product-count-level churn rate "
        f"at {highest_product_churn:.2f}%."
    )


# Identify correlation with churn
churn_correlations = correlation_matrix[
    "exited"
].drop("exited")

strongest_positive = churn_correlations.idxmax()
strongest_negative = churn_correlations.idxmin()

add_finding(
    "Correlation Analysis",
    f"The strongest positive numerical correlation with Exited "
    f"among the selected features is {strongest_positive} "
    f"({churn_correlations.max():.3f})."
)

add_finding(
    "Correlation Analysis",
    f"The strongest negative numerical correlation with Exited "
    f"among the selected features is {strongest_negative} "
    f"({churn_correlations.min():.3f})."
)


# Identify high-balance inactive customer churn
if not balance_activity_table.empty:

    if (
        "Inactive" in balance_activity_table.columns
        and "High Balance" in balance_activity_table.index
    ):

        high_balance_inactive_churn = (
            balance_activity_table.loc[
                "High Balance", "Inactive"
            ]
        )

        add_finding(
            "High-Balance Inactive Customers",
            f"High-balance inactive customers have an observed churn "
            f"rate of {high_balance_inactive_churn:.2f}%."
        )


# Identify RSI comparison
if "Retained" in customer_profile.index and "Exited" in customer_profile.index:

    retained_rsi = customer_profile.loc[
        "Retained", "Average_RSI"
    ]

    exited_rsi = customer_profile.loc[
        "Exited", "Average_RSI"
    ]

    add_finding(
        "Relationship Strength",
        f"Average RSI is {retained_rsi:.2f} for retained customers "
        f"and {exited_rsi:.2f} for exited customers."
    )


# Create EDA findings table
eda_findings_df = pd.DataFrame(eda_findings)

print("\nEDA Findings")
display(eda_findings_df)


# Save all analytical tables
for table_name, table in eda_tables.items():

    filename = f"{table_name}.csv"

    save_table(
        table,
        filename
    )


# Save EDA findings
eda_findings_path = os.path.join(
    table_folder,
    "EDA_Findings.csv"
)

eda_findings_df.to_csv(
    eda_findings_path,
    index=False
)


# Save EDA findings as a readable text report
eda_report_path = os.path.join(
    output_folder,
    "EDA_Findings_Report.txt"
)

with open(
    eda_report_path,
    "w",
    encoding="utf-8"
) as report:

    report.write(
        "CUSTOMER ENGAGEMENT AND PRODUCT UTILIZATION ANALYTICS\n"
    )

    report.write("EXPLORATORY DATA ANALYSIS FINDINGS\n\n")

    report.write(
        f"Total Customers: {len(df)}\n"
    )

    report.write(
        f"Total Columns: {len(df.columns)}\n"
    )

    report.write(
        f"Overall Churn Rate: {overall_churn_rate:.2f}%\n\n"
    )

    for index, row in eda_findings_df.iterrows():

        report.write(
            f"{index + 1}. {row['Analysis Area']}\n"
        )

        report.write(
            f"{row['Observation']}\n\n"
        )


# Save correlation matrix separately
correlation_matrix_path = os.path.join(
    table_folder,
    "Correlation_Matrix.csv"
)

correlation_matrix.to_csv(
    correlation_matrix_path
)


# Final EDA completion summary
print("\nEDA ANALYSIS COMPLETED")

print("Total Customers Analysed:", len(df))
print("Total Original Columns:", 14)
print("Total Engineered Features:", len([
    "age_group",
    "tenure_group",
    "balance_category",
    "product_count_category",
    "activity_status",
    "customer_status",
    "high_balance_flag",
    "inactive_high_balance_flag",
    "product_relationship",
    "premium_customer_candidate",
    "relationship_strength_index",
    "relationship_strength_category"
]))

print("Overall Churn Rate:", round(overall_churn_rate, 2), "%")

print("\nEDA Charts Folder:")
print(eda_figure_folder)

print("\nAnalytical Tables Folder:")
print(table_folder)

print("\nEDA Findings Report:")
print(eda_report_path)

print("\nCorrelation Matrix:")
print(correlation_matrix_path)

print("\nAll EDA charts, analytical tables and findings saved successfully.")
```
